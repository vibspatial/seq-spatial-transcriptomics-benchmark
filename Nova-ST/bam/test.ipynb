```python
import pysam
from collections import Counter

bam_path = '/data/groups/technologies/spatial.catalyst/Projects/2024-07-UTBenchmark-SpC/data/processed/Nova-ST/Exp65-sampleSPC014/NovaScope/novascope_output/align/NovaSeq6000_20241107/03D1/Exp65-sampleSPC014/bam/sttoolsAligned.sortedByCoord.out.bam'

stats = Counter()

def _valid_tag(x):
    return x not in (0, None, "", "-", ".")

with pysam.AlignmentFile(bam_path, "rb") as bam:
    for read in bam:
        if read.has_tag("sM") and read.get_tag("sM") >= 0:
            stats["sM_ge_0"] += 1
            
            if read.has_tag("GN") and read.get_tag("GN") != "-":
                stats["gene_assigned"] += 1
            
            if read.has_tag("NH") and read.get_tag("NH") == 1:
                if not read.has_tag("CB"):
                    stats["no_CB_tag"] += 1
                    continue
                elif read.get_tag("CB") == "-":
                    stats["CB_dash"] += 1

                if not read.has_tag("UB"):
                    stats["no_UB_tag"] += 1
                elif read.get_tag("UB") == "-":
                    stats["UB_dash"] += 1

                if not read.has_tag("GN"):
                    stats["no_GN_tag"] += 1
                elif read.get_tag("GN") == "-":
                    stats["GN_dash"] += 1
                    
                if read.has_tag("CB") and read.has_tag("UB") and read.has_tag("GN"):
                    stats["included_NH1"] += 1
                
            if read.has_tag("CB") and read.has_tag("UB") and read.has_tag("GN"):
                stats["included_total"] += 1
                
                if _valid_tag(read.get_tag("CB")) and _valid_tag(read.get_tag("UB")) and _valid_tag(read.get_tag("GN")):
                    stats["included_valid"] += 1
                    
            if read.is_secondary:
                stats["reads_secondary"] += 1
            
            if read.is_supplementary:
                stats["reads_supplementary"] += 1
                
        if read.is_secondary or read.is_supplementary:
            continue
        
        if read.has_tag("sM") and read.get_tag("sM") >= 0:
            stats["gpt_sM_ge_0"] += 1
            
            if read.has_tag("GN") and read.get_tag("GN") != "-":
                stats["gpt_gene_assigned"] += 1

print(stats)
```

```python
for key, value in stats.items():
    print(f"{key}: {value:,}")
```

```python
import pysam
from collections import Counter

bam_path = '/data/groups/technologies/spatial.catalyst/Projects/2024-07-UTBenchmark-SpC/data/processed/Nova-ST/Exp65-sampleSPC014/NovaScope/novascope_output/align/NovaSeq6000_20241107/03D1/Exp65-sampleSPC014/bam/sttoolsAligned.sortedByCoord.out.bam'

stats = Counter()

def _valid_tag(x):
    return x not in (0, None, "", "-", ".")

seen_reads = set()

with pysam.AlignmentFile(bam_path, "rb") as bam:
    for read in bam:
        
        if read.query_name in seen_reads:
            continue
        
        if read.has_tag("sM") and read.get_tag("sM") >= 0:
            if read.has_tag("CB") and read.has_tag("UB") and read.has_tag("GN"):
                if _valid_tag(read.get_tag("CB")) and _valid_tag(read.get_tag("UB")) and _valid_tag(read.get_tag("GN")):
                    stats["included_valid"] += 1
                    seen_reads.add(read.query_name)
                    
print(stats)
```
