## scalefactors_json.json

scalefactors_json.json: The purpose of the file is to record the relative scales of the user-supplied image, the images in the spatial outs, and the Visium array. This file may contain the following fields:
- regist_target_img_scalef: A scaling factor that converts pixel positions in the original, full-resolution microscope image to the downsampled version used in image registration.
- tissue_hires_scalef: A scaling factor that converts pixel positions in the original, full-resolution image to pixel positions in tissue_hires_image.png.
- tissue_lowres_scalef: A scaling factor that converts pixel positions in the original, full-resolution image to pixel positions in tissue_lowres_image.png.
- fiducial_diameter_fullres: The number of pixels that span the diameter of a fiducial spot in the original, full-resolution image.
- spot_diameter_fullres: The number of pixels that span the diameter of a spot in the original, full-resolution image. This field is intended for visualization purposes, and can vary for different slide designs, ranging from 60-70 µm. Spot diameters and spot positions are estimates. It is better to use the known pixel dimensions of a calibrated microscope rather than trying to infer pixel dimensions from the spot diameter.

## Apply bigwarp transformation to barcode positions

```python
import numpy as np
import pandas as pd
import h5py
import json

def save_cytassist_coordinates_for_bigwarp(tissue_positions_path, feature_slice_path, cytassist_microns_per_pixel, output_path, bin_size):
    # Load the tissue positions
    df = pd.read_parquet(tissue_positions_path)

    # Load the metadata and extract affine transformation matrix
    with h5py.File(feature_slice_path, "r") as f5:
        metadata = json.loads(dict(f5.attrs)["metadata_json"])
    affine_matrix = np.array(metadata['transform_matrices']['spot_colrow_to_cytassist_colrow'])

    # Apply the affine transformation matrix
    if bin_size == 2:
        coords = np.hstack([df[['array_col', 'array_row']].values, np.ones((df.shape[0], 1))])
    elif bin_size > 2:
        coords = np.hstack([df[['array_col', 'array_row']].values * (bin_size/2), np.ones((df.shape[0], 1))])
    if bin_size < 2:
        raise ValueError("bin_size must be 2 or greater.")
    
    transformed_coords = coords @ affine_matrix.T

    # Save the transformed coordinates
    df_transformed = pd.DataFrame({
        'x': transformed_coords[:, 0] * (cytassist_microns_per_pixel / 1000), # bigwarp needs the coordinates in millimeters.
        'y': transformed_coords[:, 1] * (cytassist_microns_per_pixel / 1000),
    })
    
    df_transformed.to_csv(output_path, index=False)
```

Use warp_points_from_csv.groovy script in FIJI to calculate transformed coordinates. This script takes a landmark.csv file made in BigWarp and allows transforming coordinates in both directions (i.e. from moving image to fixed image or from fixed image to moving image) based on a variety of different transformation types (including affine and thin-plate spline). Note that this requires the input coordinates to be in millimeters and that the output coordinates will be in millimeters as well.

```python
import pandas as pd

def convert_bigwarp_to_pixel_coordinates(tissue_positions_path, bigwarp_csv, fullres_microns_per_pixel, output_path):
    # Load the tissue positions
    df_tissue = pd.read_parquet(tissue_positions_path)
    
    # Read the CSV file containing transformed coordinates from bigwarp
    df_bigwarp = pd.read_csv(bigwarp_csv)

    # Convert coordinates to pixels
    df_pxl = pd.DataFrame({
        'bigwarp_x': df_bigwarp['x'] * ((1 / fullres_microns_per_pixel) * 1000),
        'bigwarp_y': df_bigwarp['y'] * ((1 / fullres_microns_per_pixel) * 1000),
    })

    # Add the new coordinates to the tissue positions data
    df_merged = df_tissue.join(df_pxl)

    # Calculate distance between coordinates
    df_merged['distance_in_pxl'] = np.sqrt(
        (df_merged['pxl_col_in_fullres'] - df_merged['bigwarp_x']) ** 2 +
        (df_merged['pxl_row_in_fullres'] - df_merged['bigwarp_y']) ** 2
    ) 
    df_merged['distance_in_micron'] = df_merged['distance_in_pxl'] * fullres_microns_per_pixel
    
    # Save the merged DataFrame as either CSV or parquet based on user input
    if output_path.endswith('.csv'):
        df_merged.to_csv(output_path, index=False)
    elif output_path.endswith('.parquet'):
        df_merged.to_parquet(output_path, index=False)
    else:
        raise ValueError("Invalid output format. Use '.csv' or '.parquet' as file extension.")
```
