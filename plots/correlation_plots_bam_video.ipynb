```python
import cv2
import os
from natsort import natsorted
import re
import numpy as np

# Folder containing PNG images
base_dir = "/data/groups/technologies/spatial.catalyst/Projects/2024-07-UTBenchmark-SpC/data/processed/plots/bam_plots"

for sample_type in ["Mouse_lung", "Mouse_liver", "Mouse_brain", "Human_lung"]:
    # Input image folder
    image_folder = os.path.join(base_dir, sample_type, "gene_counts")

    # Output video file
    output_video = os.path.join(image_folder, "video.mp4")

    # Frames per second
    fps = 3

    # Get list of all PNG files, naturally sorted
    images = [img for img in os.listdir(image_folder) if img.endswith(".png")]
    images = natsorted(images)

    # Read the first image to get frame size
    frame = cv2.imread(os.path.join(image_folder, images[0]))
    height, width, layers = frame.shape
    
    # Add extra space at the top for text
    top_margin = 80
    new_height = height + top_margin

    # Define the codec and create VideoWriter object
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # for .mp4
    video = cv2.VideoWriter(output_video, fourcc, fps, (width, new_height))

    for image in images:
        img_path = os.path.join(image_folder, image)
        frame = cv2.imread(img_path)
        
        new_frame = np.ones((new_height, width, 3), dtype=np.uint8) * 255
        new_frame[top_margin:, :, :] = frame
        
        match = re.search(r'sample_size_(\d+)', image)
        if match:
            sample_size = int(match.group(1))
            text = f"Sample size: {sample_size:,}"  # adds commas for readability
        else:
            text = ""
        
        font = cv2.FONT_HERSHEY_SIMPLEX
        font_scale = 1.2
        color = (0, 0, 0)  # black
        thickness = 2
        position = (50, int(top_margin / 2 + 10))
        cv2.putText(new_frame, text, position, font, font_scale, color, thickness, cv2.LINE_AA)
        
        video.write(new_frame)

    video.release()
    print("Video saved to", output_video)
```
