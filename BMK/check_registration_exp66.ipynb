```python
import gzip
import pandas as pd

barcodes = r'd:\Data\exp66\BSTViewer_project\level_matrix\level_1\barcodes_pos.tsv.gz'
tissue_barcodes = r'd:\Data\exp66\BSTViewer_project\subdata\L1_heAuto\barcodes_pos.tsv.gz'

# Open and read the .tsv.gz file
with gzip.open(barcodes, 'rt') as f:  # 'rt' mode is for reading text
    # Load the TSV file into a pandas DataFrame
    df_barcodes = pd.read_csv(f, sep='\t', header=None)

# Open and read the .tsv.gz file
with gzip.open(tissue_barcodes, 'rt') as f:  # 'rt' mode is for reading text
    # Load the TSV file into a pandas DataFrame
    df_barcodes_tissue = pd.read_csv(f, sep='\t', header=None)
```

```python
from skimage import io
from matplotlib import pyplot as plt

img = io.imread(r'd:\Data\exp66\BSTViewer_project\he.tif')
img.shape
```

```python
import numpy as np

def calculate_scaling_factor(width, height):
    std_width = 1000
    std_height = std_width / (46 * 31) * (46 * 36 * np.sqrt(3) / 2.0)
    
    if std_width / std_height > width / height:
        scale = width / std_width
    else:
        scale = height / std_height
        
    return(scale)
```

```python
scaling_factor = calculate_scaling_factor(img.shape[1], img.shape[0])
scaling_factor
```

```python
# Image
plt.figure(figsize=(25,25))
plt.imshow(img)
plt.xlim(0,img.shape[1])
plt.ylim(img.shape[0],0)
```
