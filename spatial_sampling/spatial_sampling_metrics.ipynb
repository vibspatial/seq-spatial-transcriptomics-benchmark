## Estimate missing spots

```python
import gc
import math

import numpy as np
import pandas as pd
import pyarrow.dataset as ds
import rasterio
from rasterio.transform import from_origin
import spatialdata as sd
from skimage.morphology import binary_dilation, binary_erosion, disk, remove_small_holes

from spatial_sampling_metrics import SPOT_CONFIG, get_capture_area

mask_pixel_size_um = 0.4
expansion_um = {"BMKMANU S1000": 1, "BMKMANU S3000": 0.5, "Nova-ST": 0.4, "Stereo-Seq": 0.4}
ideal_lattice_pitch_um = {
    "Stereo-Seq": 0.5,
    "BMKMANU S1000": 4.8,
    "BMKMANU S3000": 3.5,
    "Nova-ST": 0.625,
}
lattice_type = {
    "Stereo-Seq": "square",
    "BMKMANU S1000": "triangular",
    "BMKMANU S3000": "triangular",
    "Nova-ST": "triangular",
}
maximum_filled_gap_um2 = 400.0
mask_technologies = {"BMKMANU S1000", "BMKMANU S3000", "Nova-ST", "Stereo-Seq"}
mask_output_dir = output_path / "physical_spot_masks"
mask_output_dir.mkdir(parents=True, exist_ok=True)

maximum_filled_gap_pixels = math.ceil(
    maximum_filled_gap_um2 / mask_pixel_size_um**2
)
mask_statistics = []

def mark_spot_footprints(mask, x, y, min_x, max_y, pixel_size, shape, size_um):
    """Mark every raster pixel intersected by a physical spot."""
    column0 = np.floor((x - min_x) / pixel_size).astype(np.int64)
    row0 = np.floor((max_y - y) / pixel_size).astype(np.int64)
    local_x = (x - min_x) - column0 * pixel_size
    local_y = (max_y - y) - row0 * pixel_size
    zeros = np.zeros_like(local_x)
    half_size = size_um / 2
    reach = math.ceil(half_size / pixel_size)
    for row_offset in range(-reach, reach + 1):
        dy = np.maximum.reduce([
            row_offset * pixel_size - local_y,
            local_y - (row_offset + 1) * pixel_size,
            zeros,
        ])
        for column_offset in range(-reach, reach + 1):
            dx = np.maximum.reduce([
                column_offset * pixel_size - local_x,
                local_x - (column_offset + 1) * pixel_size,
                zeros,
            ])
            overlaps = (
                dx**2 + dy**2 <= half_size**2
                if shape == "circle"
                else (dx <= half_size) & (dy <= half_size)
            )
            row = row0 + row_offset
            column = column0 + column_offset
            valid = (
                overlaps & (row >= 0) & (row < mask.shape[0])
                & (column >= 0) & (column < mask.shape[1])
            )
            mask[row[valid], column[valid]] = True

for sample in sample_metadata:
    if sample["technology"] not in mask_technologies:
        continue

    sample_expansion_um = expansion_um[sample["technology"]]
    footprint_radius_pixels = math.ceil(
        sample_expansion_um / mask_pixel_size_um
    )
    footprint = disk(footprint_radius_pixels)

    technology_name = sample["technology"] + (
        f"_{sample['replicate']}" if sample.get("replicate") else ""
    )
    result_dir = output_path / sample["sample_type"] / technology_name
    metrics_path = result_dir / "spot_metrics"
    if not metrics_path.exists():
        metrics_path = result_dir / "spot_metrics.parquet"
    metrics = ds.dataset(metrics_path, format="parquet", partitioning="hive")

    sdata = sd.read_zarr(sample["sdata_path"])
    capture_area = get_capture_area(sdata, sample)
    min_x, min_y, max_x, max_y = capture_area.bounds
    width = math.ceil((max_x - min_x) / mask_pixel_size_um)
    height = math.ceil((max_y - min_y) / mask_pixel_size_um)
    transform = from_origin(
        min_x, max_y, mask_pixel_size_um, mask_pixel_size_um
    )
    occupied = np.zeros((height, width), dtype=bool)
    spot_config = SPOT_CONFIG[sample["technology"]]
    ideal_pitch_um = ideal_lattice_pitch_um[sample["technology"]]
    sample_lattice_type = lattice_type[sample["technology"]]
    ideal_voronoi_area_um2 = (
        ideal_pitch_um**2
        if sample_lattice_type == "square"
        else np.sqrt(3) / 2 * ideal_pitch_um**2
    )

    number_of_spots = 0
    scanner = metrics.scanner(
        columns=["x_um", "y_um"],
        batch_size=500_000,
    )
    for batch in scanner.to_batches():
        x = batch.column(0).to_numpy(zero_copy_only=False)
        y = batch.column(1).to_numpy(zero_copy_only=False)
        number_of_spots += len(x)
        mark_spot_footprints(
            occupied, x, y, min_x, max_y, mask_pixel_size_um,
            spot_config["shape"], spot_config["size_um"],
        )

    expanded = binary_dilation(occupied, footprint=footprint)
    filled = remove_small_holes(
        expanded,
        area_threshold=maximum_filled_gap_pixels,
        connectivity=2,
    )
    padding = footprint_radius_pixels
    padded = np.pad(filled, padding, mode="constant", constant_values=True)
    eroded = binary_erosion(padded, footprint=footprint)
    final_mask = eroded[padding:-padding, padding:-padding]

    masked_pixels = int(final_mask.sum())
    total_pixels = height * width
    unmasked_pixels = total_pixels - masked_pixels
    pixel_area_um2 = mask_pixel_size_um**2
    masked_area_mm2 = masked_pixels * pixel_area_um2 / 1_000_000
    unmasked_area_mm2 = unmasked_pixels * pixel_area_um2 / 1_000_000
    theoretical_spots = (
        masked_pixels * pixel_area_um2 / ideal_voronoi_area_um2
    )
    missing_spots_percent = 100 * (
        1 - number_of_spots / theoretical_spots
    )

    mask_values = final_mask.astype(np.uint8) * 255
    safe_name = (
        f"{sample['sample_type']}_{technology_name}_{sample['dataset_id']}"
        .replace(" ", "_").replace("/", "-").replace("'", "")
    )
    mask_path = mask_output_dir / f"{safe_name}_physical_spot_mask.tif"
    with rasterio.open(
        mask_path,
        "w",
        driver="GTiff",
        height=height,
        width=width,
        count=1,
        dtype="uint8",
        transform=transform,
        compress="deflate",
        tiled=True,
        BIGTIFF="IF_SAFER",
    ) as destination:
        destination.write(mask_values, 1)

    mask_statistics.append({
        "sample_type": sample["sample_type"],
        "technology": sample["technology"],
        "replicate": sample.get("replicate"),
        "dataset_id": sample["dataset_id"],
        "mask_file": str(mask_path),
        "pixel_size_um": mask_pixel_size_um,
        "expansion_um": sample_expansion_um,
        "masked_area_mm2": masked_area_mm2,
        "masked_area_percent": 100 * masked_pixels / total_pixels,
        "unmasked_area_mm2": unmasked_area_mm2,
        "unmasked_area_percent": 100 * unmasked_pixels / total_pixels,
        "number_of_spots": number_of_spots,
        "lattice_type": sample_lattice_type,
        "ideal_lattice_pitch_um": ideal_pitch_um,
        "ideal_voronoi_area_um2": ideal_voronoi_area_um2,
        "theoretical_physical_spots": theoretical_spots,
        "missing_spots_percent": missing_spots_percent,
    })
    print(f"Saved {mask_path}")
    del sdata, occupied, expanded, filled, padded, eroded, final_mask, mask_values
    gc.collect()

mask_statistics = pd.DataFrame(mask_statistics)
mask_statistics_path = mask_output_dir / "physical_spot_mask_statistics.csv"
mask_statistics.to_csv(mask_statistics_path, index=False)
print(f"Saved {mask_statistics_path}")
mask_statistics
```
