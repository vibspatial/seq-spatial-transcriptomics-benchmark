## Switch moving and fixed in Bigwarp landmarks csv

```python
import csv

# Input and output file paths
input_file = r'e:\2024-07-UTBenchmark-SpC\data\processed\Visium_HD\1\SCA002_reseq\registration_tests\Bigwarp\4_cytassist_to_slidescanner_KH\landmarks_KH.csv'
output_file = r'e:\2024-07-UTBenchmark-SpC\data\processed\Visium_HD\1\SCA002_reseq\registration_tests\Bigwarp\4_cytassist_to_slidescanner_KH\landmarks_KH_swapped.csv'

# Open the input file for reading and the output file for writing
with open(input_file, 'r', newline='') as csvfile_in, open(output_file, 'w', newline='') as csvfile_out:
    reader = csv.reader(csvfile_in, quotechar='"')
    writer = csv.writer(csvfile_out, quotechar='"', quoting=csv.QUOTE_ALL)

    # Iterate through each row in the input CSV
    for row in reader:
        # Swap columns
        row[2], row[4] = row[4], row[2]
        row[3], row[5] = row[5], row[3]

        # Write the modified row to the output file
        writer.writerow(row)
```
